```python
import numpy as np

# 1. Create a numpy array of size 2 x 2
arr1 = np.array([[1, 2], [3, 4]])
print("2 x 2 array:")
print(arr1)

# 2. Create a random numpy array of size 3 x 4
arr2 = np.random.rand(3, 4)
print("\n3 x 4 random array:")
print(arr2)

# 3. Create a numpy array of ones and a numpy array of zeroes
arr_ones = np.ones((3, 3))
arr_zeros = np.zeros((2, 2))
print("\nOnes array:")
print(arr_ones)
print("Zeros array:")
print(arr_zeros)

# 4. Create a random numpy array of size 6x3 and transpose it
arr4 = np.random.rand(6, 3)
arr4_t = arr4.transpose()
print("\nOriginal 6x3 array:")
print(arr4)
print("Transposed array:")
print(arr4_t)

# 5. Create two random numpy array of size n x p and p x m and then compute the dot product
n, p, m = 2, 3, 4
A = np.random.rand(n, p)
B = np.random.rand(p, m)
dot_result = A @ B
print("\nMatrix A ({}x{}):".format(n, p))
print(A)
print("Matrix B ({}x{}):".format(p, m))
print(B)
print("Dot product result ({}x{}):".format(n, m))
print(dot_result)
```
